```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
xs = np.linspace(0,1,100)
plt.plot(xs, xs**2)
plt.fill_between(xs, xs**2, alpha=0.3)
plt.title('f(x)=x^2 在[0,1]上的面积')
plt.show()
```
